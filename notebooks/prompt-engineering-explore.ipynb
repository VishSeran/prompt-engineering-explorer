```python
model_name = "meta-llama/Llama-3.2-1B-Instruct"
```

```python
llm = LLM(model=model_name,  gpu_memory_utilization=0.7)
```

```python
def get_llm_response (prompt, params=None):
    
    default_params = SamplingParams(
        max_tokens=256,
        min_tokens=128,
        temperature=0.7,
        top_p = 0.8,
        top_k = 40
    )
    
    if params:
        default_params = params
        
    response = llm.generate(prompt, default_params)
```
