```python
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
GEMINI_KEY = os.environ.get("GEMINI_KEY")
```

### Zero-Shot Prompting

```python
def llm_reponse_zero_shot ():
    
    prompt_template = PromptTemplate(
        input_variables=["sentence"],
        template = """Classify the sentiment of the following review as Positive, Negative, or Neutral.
Only respond with one word.

Review: {sentence}
Sentiment:"""
    )
    
    llm = ChatGoogleGenerativeAI(
        model="gemini-3-flash-preview",
        temperature=0.3,
        api_key = GEMINI_KEY)
    
    zero_shot_chain = prompt_template | llm | StrOutputParser()
    return zero_shot_chain
```

```python
reviews = [
    "The battery life on this phone is absolutely terrible.",
    "Delivery was on time and the packaging was neat.",
    "It works, I guess. Nothing special about it."
]
```

```python
zero_shot = llm_reponse_zero_shot()
```

```python
for r in reviews:
    response = zero_shot.invoke({"sentence": r})
    print(f"Review : {r}")
    print(f"Result : {response.strip()}\n")
```

#### One-Shot Learning

```python
def one_shot_prompting ():
    
    llm = ChatGoogleGenerativeAI(
        model = "gemini-3-flash-preview",
        temperature = 0.7,
        api_key = GEMINI_KEY
    )
    
    prompt_template = PromptTemplate(
        input_variables= ["sentence"],
        template= """
        Translate the sentence from English to French.
        
        Example:
        English: How is the weather today?
        French: Comment est le temps aujourd'hui?
        
        Now translate this,
        English: {sentence}
        French:"""
    )
    
    one_shot_chain = prompt_template | llm | StrOutputParser()
    return one_shot_chain
```

```python
sentences = ["Where is the nearest supermarket?",
    "I would like a table for two, please.",]

one_shot_chain = one_shot_prompting()

for sentence in sentences:
    response = one_shot_chain.invoke({"sentence": sentence})
    print("Input: ", sentence)
    print("Response: ", response,"\n")
```

#### Few-Shot Prompting

```python
examples = [
    {"text": "URGENT: Your account has been SUSPENDED! Click here NOW!",
     "tone": "Aggressive"},
    {"text": "Hey! Just checking in — hope you're having a great week.",
     "tone": "Friendly"},
    {"text": "Please find attached the Q3 financial report for your review.",
     "tone": "Formal"},
    {"text": "Ugh, I can't believe this happened AGAIN. So frustrating.",
     "tone": "Negative"},
]
```

```python
def few_shot_prompting ():
    
    llm = ChatGoogleGenerativeAI(
        model = "gemini-3-flash-preview",
        temperature = 0.7,
        api_key = GEMINI_KEY
    )
    
    promt_tempate = PromptTemplate(
        input_variables=["text", "tone"],
        template="""Text: {text}, Tone: {tone}"""
    )
    
    few_shot_prompt = FewShotPromptTemplate(
        examples = examples,
        example_prompt = promt_tempate,
        prefix = "Classify the tone of each text as: Friendly, Formal, Aggressive, or Negative.\n",
        suffix = "Text:{text}, Tone:",
        input_variables = ["text"]
    )
    
    few_shot_chain = few_shot_prompt | llm | StrOutputParser()
    
    return few_shot_chain
```

```python
test_texts = [
    "We regret to inform you that your application was unsuccessful.",
    "This is COMPLETELY unacceptable. I want a refund immediately!",
    "Hi there! Would love to grab coffee sometime soon!",
]
```

```python
few_shot_chain = few_shot_prompting()

for text in test_texts:
    response = few_shot_chain.invoke({"text":text})
    print("Input: ", text)
    print("Response: ",response, "\n")
```
