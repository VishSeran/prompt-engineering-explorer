```python
from langchain_openai import ChatOpenAI
from langchain_ollama import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
GEMINI_KEY = os.environ.get("GEMINI_KEY")
```

### Zero-Shot Prompting

```python
def llm_reponse_zero_shot ():
    
    prompt_template = PromptTemplate(
        input_variables=["sentence"],
        template = """Classify the sentiment of the following review as Positive, Negative, or Neutral.
Only respond with one word.

Review: {sentence}
Sentiment:"""
    )
    
    llm = ChatGoogleGenerativeAI(
        model="gemini-3-flash-preview",
        temperature=0.3,
        api_key = GEMINI_KEY)
    
    zero_shot_chain = prompt_template | llm | StrOutputParser()
    return zero_shot_chain
```

```python
reviews = [
    "The battery life on this phone is absolutely terrible.",
    "Delivery was on time and the packaging was neat.",
    "It works, I guess. Nothing special about it."
]
```

```python
zero_shot = llm_reponse_zero_shot()
```

```python
for r in reviews:
    response = zero_shot.invoke({"sentence": r})
    print(f"Review : {r}")
    print(f"Result : {response.strip()}\n")
```

#### One-Shot Learning

```python
def one_shot_prompting ():
    
    llm = ChatGoogleGenerativeAI(
        model = "gemini-3-flash-preview",
        temperature = 0.7,
        api_key = GEMINI_KEY
    )
    
    prompt_template = PromptTemplate(
        input_variables= ["sentence"],
        template= """
        Translate the sentence from English to French.
        
        Example:
        English: How is the weather today?
        French: Comment est le temps aujourd'hui?
        
        Now translate this,
        English: {sentence}
        French:"""
    )
    
    one_shot_chain = prompt_template | llm | StrOutputParser()
    return one_shot_chain
```

```python
sentences = ["Where is the nearest supermarket?",
    "I would like a table for two, please.",]

one_shot_chain = one_shot_prompting()

for sentence in sentences:
    response = one_shot_chain.invoke({"sentence": sentence})
    print("Input: ", sentence)
    print("Response: ", response,"\n")
```
